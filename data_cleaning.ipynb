# Datos mensuales

```python
def clean_monthly_data(file_path, sheet_name, skiprows):
    # Leer los datos desde un archivo Excel, especificando la hoja, las filas a saltar y las columnas a usar
    df = pd.read_excel(file_path, sheet_name=sheet_name, skiprows=skiprows, usecols='A:M', nrows=45)

    # Transformar el DataFrame de formato ancho a largo
    # 'AÑOS' se mantiene como identificador, y las columnas restantes se convierten en filas
    df = df.melt(id_vars=['AÑOS'], var_name='Mes', value_name='Valor')

    # Diccionario para mapear nombres de meses a números
    meses = {
        'ENE': 1, 'FEB': 2, 'MAR': 3, 'ABR': 4, 'MAY': 5, 'JUN': 6,
        'JUL': 7, 'AGO': 8, 'SEP': 9, 'OCT': 10, 'NOV': 11, 'DIC': 12
    }

    # Mapear los nombres de los meses a números utilizando el diccionario
    df['Mes_num'] = df['Mes'].map(meses)

    # Renombrar las columnas para un formato más estándar
    df = df.rename(columns={'AÑOS': 'year', 'Mes_num': 'month'})
    
    # Crear una columna de fecha a partir del año y el mes, estableciendo el primer día del mes como día
    df['date'] = pd.to_datetime(df[['year', 'month']].assign(day=1))

    # Seleccionar solo las columnas de fecha y valor
    #df = df[['date', 'Valor']]
    
    df = df.dropna()
    
    # Ordenar el DataFrame por la columna de fecha
    df = df.sort_values('date')
    
    return df
```

```python
file_path = 'data/DATOS METEREOLOGICOS MENSUALES.xlsx'
sheet_name = 'PRECIPITACION TOTAL MENSUAL'

df_precip_monthly = clean_monthly_data(file_path, sheet_name, 12)

csv_file_path = 'data/precipitaciones_mensual.csv'
df_precip_monthly.to_csv(csv_file_path, index=False)
```
